## 📈 Statistical Parametric Mapping (SPM)

Statistical parametric mapping (SPM) uses Random Field Theory to make statistical inferences about normalized sets of biomechanical variables across time (1D). It can be used to compare kinematic, kinetic, or EMG time-series curves using traditional NHST statistical tests such as a t-test, ANOVA, and linear regression. In this course, we will be using the package, `spm1d`, in [Python](https://spm1d.org/index.html) to perform some basic SPM analyses.

Before we perform the SPM analyses, we must prepare the time-series data so that they can be properly structured for SPM. The process of preparing and organizing raw data into usable data is known as **data wrangling**. For SPM, the time-series data must be structured so that each variable for each condition (i.e., conventional, sumo) is its own data file, where the data is organized in a *JxQ* array:

* J = number of trials or subjects
* Q = number of frames or time points (i.e., 100)

The time-series data are averaged across subjects or trials at each frame in a process known as **ensemble averaging** where the solid thick lines represent the mean and the semi-transparent bands represent the +/- 1 standard deviation:

<img src="https://raw.githubusercontent.com/plnu-biomechanics/kin6015/main/notebooks/images/SPM_plot_sample.png">

Rather than simply comparing a discrete point such as maximum knee flexion, for example, the entire knee flexion-extension angle can be examined between conditions or groups (e.g., training vs. carbon-fiber, male vs. female, etc.).

### 🔄 Data Wrangling: Restructuring to $J \times Q$ Arrays for SPM

For `spm1d` analyses, the time-series data must be formatted as a $J \times Q$ array (where rows $J$ are subjects/trials, and columns $Q$ are normalized percentage timepoints or frames, here $101$ frames). We will pivot our `running_df` to create these arrays for each variable (`knee_flexion`, `ankle_dorsiflexion`, `foot_rotation`) and each shoe condition (`training`, `carbonfiber`).

````python
# =======================================
#  Code generated or assisted by Colab
# =======================================
import numpy as np
import pandas as pd
import os

# Variables to extract
kinematic_vars = ['knee_flexion', 'ankle_dorsiflexion', 'foot_rotation']
conditions = ['training', 'carbonfiber']

print("=== Restructuring Data into JxQ Arrays ===\n")

for var in kinematic_vars:
    for cond in conditions:
        # Filter by condition
        sub_df = running_df[running_df['group'] == cond]

        # Pivot to obtain rows as players (J) and columns as frames (Q)
        # Index is 'player_id', Columns is 'frame', Values is the kinematic variable
        jq_df = sub_df.pivot(index='player_id', columns='frame', values=var)

        # Convert to numpy array
        jq_array = jq_df.to_numpy()

        # Verify shape
        print(f"Variable: {var} | Condition: {cond}")
        print(f"  - Players (J): {jq_array.shape[0]}")
        print(f"  - Frames (Q): {jq_array.shape[1]}")
        print(f"  - Array Shape: {jq_array.shape}")

        # Export to a text file for spm1d
        output_filename = f"{var}_{cond}_JxQ.txt"
        output_path = os.path.join(data_dir, output_filename)
        np.savetxt(output_path, jq_array, fmt='%.5f')
        print(f"  - Saved to: {output_path}\n")
````

````python
# =======================================
#  Code generated or assisted by Colab
# =======================================

!pip install spm1d -q

import os
import numpy as np
import matplotlib.pyplot as plt
import spm1d

# Configure plotting style
plt.rcParams.update({'font.size': 11})
fig, axes = plt.subplots(1, 3, figsize=(16, 5))

kinematic_vars = ['knee_flexion', 'ankle_dorsiflexion', 'foot_rotation']
titles = ['Knee Flexion', 'Ankle Dorsiflexion', 'Foot Rotation']
data_dir = "kin4042/lab5"

total_frames = 101
num_ticks = 6  # Number of ticks (0%, 20%, ..., 100%)

def plot_spm(ax, Y1, Y2, label_Y1, label_Y2, x_label, y_label, y_limits, show_legend):
    spm1d.plot.plot_mean_sd(Y1, label=label_Y1, linecolor='k', facecolor='k', ax=ax, lw=1.5, alpha=0.3)
    spm1d.plot.plot_mean_sd(Y2, label=label_Y2, linecolor='r', facecolor='r', linestyle='--', lw=1.5, alpha=0.20, ax=ax)
    ax.axhline(y=0, color='k', linestyle=':')

    # Conduct t-test
    alpha = 0.05
    t = spm1d.stats.ttest2(Y1, Y2, equal_var=False)
    ti = t.inference(alpha, two_tailed=True, interp=True)

    # Highlight suprathreshold regions
    for cluster in ti.clusters:
        start, end = cluster.endpoints
        ax.axvspan(start, end, ymin=0, ymax=1.0, color='grey', alpha=0.20)

    # Generate and set x-ticks
    percentages = np.linspace(0, 100, num_ticks)
    x_ticks = (percentages / 100) * (total_frames - 1)
    ax.set_xticks(x_ticks)
    ax.set_xticklabels(percentages.astype(int), fontsize=8)

    # Vertical TO line
    ax.vlines(x=40, ymin=y_limits[0]-10, ymax=y_limits[1]+10, color='k', linestyle='--', lw=1)
    ypos = y_limits[1] + (y_limits[1] - y_limits[0]) * 0.02
    ax.text(38, ypos, 'TO', fontsize=8)

    ax.set_xlabel(x_label, fontsize=10, fontweight='bold')
    ax.set_ylabel(y_label, fontsize=10, fontweight='bold')
    ax.set_ylim(y_limits)
    ax.set_xlim(0, 100)

    if show_legend:
        ax.legend(fontsize=8, loc='lower right', frameon=False)

for idx, var in enumerate(kinematic_vars):
    # Load the JxQ arrays
    trn_path = os.path.join(data_dir, f"{var}_training_JxQ.txt")
    cfp_path = os.path.join(data_dir, f"{var}_carbonfiber_JxQ.txt")

    Y_trn = np.loadtxt(trn_path)
    Y_cfp = np.loadtxt(cfp_path)

    ax = axes[idx]

    # Customize axis labels based on variable
    xlabel = "% Gait Cycle (RHS to RHS)"
    ylabel = f"{titles[idx]} (deg)"

    # Calculate shared y-axis limits
    min_val = min(Y_trn.min(), Y_cfp.min())
    if min_val > 0:
        min_val = 0
    max_val = max(Y_trn.max(), Y_cfp.max())
    y_limits = [min_val - 2, max_val + 2]

    # Plot on the pre-created subplots instead of re-creating them
    plot_spm(
        ax, Y_cfp, Y_trn,
        label_Y1="carbonfiber",
        label_Y2="training",
        x_label=xlabel,
        y_label=ylabel,
        y_limits=y_limits,
        show_legend=(idx == 0)  # Only show legend on first plot
    )

plt.tight_layout()
plt.show()
````
